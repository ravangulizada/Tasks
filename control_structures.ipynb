# Grading System:

```python
exam_marks = int(input('Input your exam marks: '))
if exam_marks >= 90:
    print('A')
elif exam_marks >= 75:
    print('B')
elif exam_marks >= 50:
    print('C')
else: 
    print('F')
```

# Multiplication Table (for loop):

```python
number = int(input('Enter number: '))
for i in range ( 1, 11 ):
    print(f'{number} x {i} = {number*i}')
```

# Password Retry System (while loop)

```python
password = 'test123'
cehd_sayi = 0
while cehd_sayi < 3:
    cehd = input('Enter the password: ')
    if cehd == password :
        print('Password is correct. ')
        break
    else: 
        cehd_sayi += 1
        if cehd_sayi == 3:
            print('You have run out of attempts. ')
        else:
            print('Password is incorrect, please try again. ')
```

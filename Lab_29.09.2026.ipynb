# Task 1

```python
score = int(input('Please enter the score: '))
if 91 <= score <= 100:
    print('Grade is A')
elif 81 <= score <= 90:
    print('Grade is B')
elif 71 <= score <= 80:
    print('Grade is C')
elif 61 <= score <= 70:
    print('Grade is D')
elif 51 <= score <= 60:
    print('Grade is E')
elif 0 <= score <= 50:
    print('Grade is F')
else:
    print(score, 'is not a score')
```

# Task 2

```python
number = int(input('Enter the number: '))
if number % 15 == 0:
    print(3, 5)
elif number % 5 == 0:
    print(5)
elif number % 3 == 0:
    print(3)
else:
    print(number)

s = 0
r = 0
for i in range (1,number+1):
    if i % 2 == 0:
        s += i
    if i % 3 == 0:
        r += 1
print('The sum of all even numbers from 1 to', number, 'is', s)
print('The number of numbers which were divisible by 3 is', r)
```
